```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
data='/content/fds2.csv'
df=pd.read_csv(data)
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df['Sales'].fillna(df['Sales'].mean(),inplace=True)
print(df.isnull().sum())
df.dropna(subset=['Sales','Product','Quantity','Region'],inplace=True)
print(df.isnull().sum())
```

```python
ps=df.groupby('Product').agg({
    'Sales':'sum',
    'Quantity':'sum'
}).reset_index()
print(ps)
```

```python
plt.figure(figsize=(10,6))
plt.bar(ps['Product'],ps['Sales'])
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Sales by Product')
plt.show()
```

```python
df['Date']=pd.to_datetime(df['Date'])
s=df.groupby('Date').agg({
    'Sales':'sum'
}).reset_index()
plt.figure(figsize=(10,6))
plt.plot(s['Date'],s['Sales'])
plt.title('Sales Over Time')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.show()
```

```python
p=df.pivot_table(values='Sales',index='Region',columns='Product',aggfunc=np.sum,fill_value=0)
print(p)
```

```python
c=df.corr(numeric_only=True)
print(c)
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(c,annot=True,cmap='coolwarm')
plt.title('CORRELATION')
plt.show()
```
