```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    'Year': list(range(2010,2021)),
    'Job posting':[150,300,350,450,550,650,750,850,950,960,990]
}
df=pd.DataFrame(data)
plt.plot(df['Year'],df['Job posting'],marker='s')
plt.title("Trend of data science job posting")
plt.xlabel("Year")
plt.ylabel("Job posting")
plt.show()
```
