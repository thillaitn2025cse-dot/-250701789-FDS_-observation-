```python
from cryptography.fernet import Fernet
key=Fernet.generate_key()
f=Fernet(key)

o=b"this is a message"
print("original data : ",o.decode())

e=f.encrypt(o)
d=f.decrypt(e)

print("encrypt data : ",e);
print("decrypt data : ",d.decode())
```
